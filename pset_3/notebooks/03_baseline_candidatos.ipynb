```python
from src import candidatos_avanzados, candidatos_clasicos

tracking.configurar_mlflow()
X, y, g = datos.separar_xy(datos.cargar_pares("train"))
todos = {**candidatos_clasicos.candidatos(), **candidatos_avanzados.candidatos()}

resultados = {}
for nombre, c in todos.items():
    fase = "baseline" if c["familia"] in ("dummy", "regla") else "candidato"
    _, resultados[nombre] = tracking.registrar_corrida(
        pipeline.modelo(c["estimador"]), nombre, c["familia"], fase, X, y, g
    )

import pandas as pd
pd.DataFrame(resultados).T.sort_values("cv_ap_mean", ascending=False)
```
